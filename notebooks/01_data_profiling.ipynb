```python
# Make a copy so the original raw data remains unchanged
sales_clean = sales.copy()
sku_master_clean = sku_master.copy()
calendar_clean = calendar.copy()
inventory_clean = inventory.copy()

# 1. Convert date columns to datetime
sales_clean['date'] = pd.to_datetime(sales_clean['date'])
sku_master_clean['launch_date'] = pd.to_datetime(sku_master_clean['launch_date'])
calendar_clean['date'] = pd.to_datetime(calendar_clean['date'])
inventory_clean['date'] = pd.to_datetime(inventory_clean['date'])

# 2. Standardize SKU IDs to uppercase
sales_clean['sku_id'] = sales_clean['sku_id'].str.upper()
sku_master_clean['sku_id'] = sku_master_clean['sku_id'].str.upper()
inventory_clean['sku_id'] = inventory_clean['sku_id'].str.upper()

# 3. Remove accidental leading/trailing spaces from text fields
sku_master_clean['category'] = sku_master_clean['category'].str.strip()
sku_master_clean['subcategory'] = sku_master_clean['subcategory'].str.strip()

print("Cleaning transformations completed.")
print()
print("Sales date dtype:", sales_clean['date'].dtype)
print("SKU launch date dtype:", sku_master_clean['launch_date'].dtype)
print("Calendar date dtype:", calendar_clean['date'].dtype)
print("Inventory date dtype:", inventory_clean['date'].dtype)

print("\nUnique Sales SKUs:", sales_clean['sku_id'].nunique())
print("Unique Master SKUs:", sku_master_clean['sku_id'].nunique())
print("Unique Inventory SKUs:", inventory_clean['sku_id'].nunique())
```
