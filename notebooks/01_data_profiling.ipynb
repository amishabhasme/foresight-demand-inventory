```python
# Core libraries for data profiling

import pandas as pd
import numpy as np

print("Libraries imported successfully.")
```

```python
# Define paths for the four FORESIGHT datasets

sales_path = "../data/raw/sales_daily.csv"
sku_master_path = "../data/raw/sku_master.csv"
calendar_path = "../data/raw/calendar.csv"
inventory_path = "../data/raw/inventory_snapshots.csv"

print("Dataset paths defined successfully.")
```

```python
# Load the FORESIGHT analysis-ready dataset

file_path = "../data/processed/analysis_ready.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully.")
print("Shape:", df.shape)
```

```python
# Check the basic structure of the dataset

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())
```

```python
# Check data types and non-null values

df.info()
```

```python
# Display the first 5 records

df.head()
```

```python
# Check missing values in every column

missing_values = df.isnull().sum()

missing_values = (
    missing_values[missing_values > 0]
    .sort_values(ascending=False)
)

missing_values
```

```python
# Calculate missing-value percentage

missing_percentage = (
    df.isnull().sum()
    / len(df)
    * 100
)

missing_summary = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": missing_percentage
})

missing_summary = (
    missing_summary[missing_summary["Missing_Count"] > 0]
    .sort_values("Missing_Count", ascending=False)
)

missing_summary
```

```python
# Check for completely duplicated records

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
# Display data types of all columns

dtype_summary = pd.DataFrame({
    "Column": df.columns,
    "Data_Type": df.dtypes.astype(str).values
})

dtype_summary
```

```python
# Identify numerical columns

numerical_columns = df.select_dtypes(
    include=np.number
).columns.tolist()

print("Numerical columns:")
for column in numerical_columns:
    print("-", column)
```

```python
# Generate descriptive statistics for numerical columns

df[numerical_columns].describe().T
```

```python
# Check the number of unique values in each column

unique_summary = pd.DataFrame({
    "Column": df.columns,
    "Unique_Values": [
        df[column].nunique(dropna=True)
        for column in df.columns
    ]
})

unique_summary
```

```python
# Check negative values in numerical columns

negative_summary = {}

for column in numerical_columns:
    negative_summary[column] = (df[column] < 0).sum()

negative_summary = (
    pd.Series(negative_summary, name="Negative_Count")
    .sort_values(ascending=False)
)

negative_summary
```

```python
# Check the date column

print("Date column data type:", df["date"].dtype)

print("\nStart date:", df["date"].min())
print("End date:", df["date"].max())

print("\nMissing dates:", df["date"].isnull().sum())
```

```python
# Convert date column to datetime for reliable date analysis

df["date"] = pd.to_datetime(
    df["date"],
    errors="coerce"
)

print("Date conversion completed.")
print("Date data type:", df["date"].dtype)
```

```python
# Check for invalid dates after conversion

invalid_dates = df["date"].isna().sum()

print("Invalid date values:", invalid_dates)

print("\nDate range:")
print("Start:", df["date"].min())
print("End:", df["date"].max())
```

```python
# Check SKU coverage

print("Total SKU records:", len(df["sku_id"]))
print("Unique SKUs:", df["sku_id"].nunique())
print("Missing SKU IDs:", df["sku_id"].isna().sum())
```

```python
# Check for whitespace and inconsistent SKU formatting

sku_as_string = df["sku_id"].astype(str)

leading_or_trailing_spaces = (
    sku_as_string != sku_as_string.str.strip()
).sum()

lowercase_skus = (
    sku_as_string != sku_as_string.str.upper()
).sum()

print("SKUs with leading/trailing spaces:", leading_or_trailing_spaces)
print("SKUs containing lowercase characters:", lowercase_skus)
```

```python
# Check revenue consistency

revenue_check = df["revenue_consistency_flag"].value_counts(
    dropna=False
)

print("Revenue consistency status:")
print(revenue_check)
```

```python
# Summarize revenue differences

revenue_difference_summary = df["revenue_difference"].describe()

revenue_difference_summary
```

```python
# Display records where revenue does not match calculated revenue

revenue_mismatches = df[
    df["revenue_consistency_flag"] == "REVENUE_MISMATCH"
]

print("Number of revenue mismatch records:",
      len(revenue_mismatches))

revenue_mismatches[
    [
        "date",
        "sku_id",
        "units_sold",
        "price",
        "revenue",
        "calculated_revenue",
        "revenue_difference"
    ]
].head(20)
```

```python
# Check inventory-related fields

inventory_columns = [
    "on_hand_units",
    "on_order_units"
]

inventory_quality = pd.DataFrame({
    "Missing_Count": df[inventory_columns].isnull().sum(),
    "Negative_Count": (df[inventory_columns] < 0).sum(),
    "Unique_Values": df[inventory_columns].nunique()
})

inventory_quality
```

```python
# Create a consolidated quality summary for key business fields

quality_summary = pd.DataFrame({
    "Metric": [
        "Total Records",
        "Unique SKUs",
        "Missing Units Sold",
        "Missing Revenue",
        "Missing Price",
        "Missing On-Hand Units",
        "Missing On-Order Units",
        "Duplicate Rows",
        "Revenue Mismatch Records"
    ],
    "Count": [
        len(df),
        df["sku_id"].nunique(),
        df["units_sold"].isnull().sum(),
        df["revenue"].isnull().sum(),
        df["price"].isnull().sum(),
        df["on_hand_units"].isnull().sum(),
        df["on_order_units"].isnull().sum(),
        df.duplicated().sum(),
        (df["revenue_consistency_flag"] == "REVENUE_MISMATCH").sum()
    ]
})

quality_summary
```

```python
# Inspect promotional information

promo_columns = [
    column for column in df.columns
    if any(keyword in column.lower()
           for keyword in ["promo", "promotion"])
]

print("Promotional columns:")
for column in promo_columns:
    print("-", column)
```

```python
# Check the values of promotional fields

for column in promo_columns:
    print(f"\n--- {column} ---")
    print(df[column].value_counts(dropna=False))
```

```python
# Identify categorical/object columns

categorical_columns = df.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("Categorical columns:")
for column in categorical_columns:
    print("-", column)
```

```python
# Check the values of promotional fields

for column in promo_columns:
    print(f"\n--- {column} ---")
    print(df[column].value_counts(dropna=False))
```

```python
# Identify categorical/object columns

categorical_columns = df.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("Categorical columns:")
for column in categorical_columns:
    print("-", column)
```

```python
# Display the most common values for categorical columns

for column in categorical_columns:
    print(f"\n--- {column} ---")
    print(df[column].value_counts(dropna=False).head(10))
```

```python
# Create a final profiling summary

profiling_summary = pd.DataFrame({
    "Metric": [
        "Total Records",
        "Total Columns",
        "Unique SKUs",
        "Start Date",
        "End Date",
        "Duplicate Rows",
        "Total Missing Values",
        "Revenue Mismatch Records"
    ],
    "Value": [
        len(df),
        len(df.columns),
        df["sku_id"].nunique(),
        df["date"].min(),
        df["date"].max(),
        df.duplicated().sum(),
        df.isnull().sum().sum(),
        (df["revenue_consistency_flag"] == "REVENUE_MISMATCH").sum()
    ]
})

profiling_summary
```

```python
# Create a clean missing-value report for documentation

missing_report = (
    df.isnull()
    .sum()
    .reset_index()
)

missing_report.columns = ["Column", "Missing_Count"]

missing_report["Missing_Percentage"] = (
    missing_report["Missing_Count"] / len(df) * 100
)

missing_report = (
    missing_report[missing_report["Missing_Count"] > 0]
    .sort_values("Missing_Count", ascending=False)
)

missing_report
```

```python
# Create a duplicate-record report

duplicate_report = pd.DataFrame({
    "Metric": ["Total Duplicate Rows"],
    "Count": [df.duplicated().sum()]
})

duplicate_report
```

```python
# Create a revenue consistency report

revenue_report = pd.DataFrame({
    "Status": df["revenue_consistency_flag"].value_counts(
        dropna=False
    ).index,
    "Record_Count": df["revenue_consistency_flag"].value_counts(
        dropna=False
    ).values
})

revenue_report["Percentage"] = (
    revenue_report["Record_Count"] / len(df) * 100
)

revenue_report
```
